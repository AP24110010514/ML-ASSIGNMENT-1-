A bag-of-words vector represents a sentence as an array of word counts, one count per vocabulary word , this ignores word order but captures what words appear and how often

```python
def bag_of_words_vector(tokens, vocabulary):
    vector = np.array([np.sum(np.array(tokens) == word) for word in vocabulary])
    return vector

vec = bag_of_words_vector(tokens, vocabulary)
print(vocabulary)
print(vec)
```

Step 4: Comparing Two Sentences (Cosine Similarity)
$$\text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|}$$


```python
sentence_a = "machine learning uses numpy arrays"
sentence_b = "deep learning also uses numpy arrays heavily"
tokens_a = sentence_a.lower().split()
tokens_b = sentence_b.lower().split()
combined_vocab = np.unique(np.array(tokens_a + tokens_b))
vec_a = bag_of_words_vector(tokens_a, combined_vocab)
vec_b = bag_of_words_vector(tokens_b, combined_vocab)
print("Vector A:", vec_a)
print("Vector B:", vec_b)
cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))
print("Cosine similarity:", cosine_sim)
```

```python
import numpy as np
import re
text = """I'm Vinuthna. I'm studying BTech. I study at SRM University.
It is located in AP."""
tokens = re.findall(r'\b\w+\b', text.lower())
print("Tokens:", tokens)
vocabulary = sorted(set(tokens))

def bag_of_words_vector(tokens, vocabulary):
    vector = np.zeros(len(vocabulary), dtype=int)

    for word in tokens:
        index = vocabulary.index(word)
        vector[index] += 1

    return vector

word_count_vector = bag_of_words_vector(tokens, vocabulary)

print("\nVocabulary:", vocabulary)
print("Word-count vector:", word_count_vector)

sorted_indices = np.argsort(word_count_vector)[::-1]
top_3_indices = sorted_indices[:3]

print("\nTop 3 most frequent words:")
for index in top_3_indices:
    print(vocabulary[index], ":", word_count_vector[index])
```

1. Takes **three** short sentences of your choice (write them as strings).
2. Tokenizes all three and builds one shared vocabulary across all of them.
3. Converts each sentence into a bag-of-words vector using that shared vocabulary.
4. Computes cosine similarity between every pair of sentences (3 pairs total: 1-2, 1-3, 2-3).
5. Prints which two sentences are the **most similar** to each other, based on the highest cosine similarity score.


```python
import numpy as np
import re

sentences = [
    "I am Vinuthna and I am studying BTech.",
    "I am studying BTech at SRM University.",
    "SRM University is located in AP."
]

tokenized = []

for sentence in sentences:
    tokens = re.findall(r'\b\w+\b', sentence.lower())
    tokenized.append(tokens)

vocabulary = sorted(set(word for tokens in tokenized for word in tokens))

def bag_of_words_vector(tokens, vocabulary):
    vector = np.zeros(len(vocabulary))
    for word in tokens:
        index = vocabulary.index(word)
        vector[index] += 1
    return vector

vectors = []

for tokens in tokenized:
    vectors.append(bag_of_words_vector(tokens, vocabulary))

def cosine_similarity(A, B):
    return np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

similarity_12 = cosine_similarity(vectors[0], vectors[1])
similarity_13 = cosine_similarity(vectors[0], vectors[2])
similarity_23 = cosine_similarity(vectors[1], vectors[2])

print("Sentence 1 & Sentence 2:", similarity_12)
print("Sentence 1 & Sentence 3:", similarity_13)
print("Sentence 2 & Sentence 3:", similarity_23)

similarities = {
    "Sentence 1 & Sentence 2": similarity_12,
    "Sentence 1 & Sentence 3": similarity_13,
    "Sentence 2 & Sentence 3": similarity_23
}

most_similar = max(similarities, key=similarities.get)

print("Most similar:", most_similar)
print("Similarity score:", similarities[most_similar])
```
